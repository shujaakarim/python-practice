### Question 2: Grades

Read the task, then run the worked solution in the next cell.

```python
marks = 80

if marks >= 80:
    print("A")
elif marks >= 65:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")

# Boundary tests requested in the question.
for marks in [80, 79, 50, 49]:
    if marks >= 80:
        grade = "A"
    elif marks >= 65:
        grade = "B"
    elif marks >= 50:
        grade = "C"
    else:
        grade = "Fail"

    print(f"{marks} marks: {grade}")
```

### Question 4: Shipping rules

Read the task, then run the worked solution in the next cell.

```python
# Test different cities, weights and order totals.
shipping_tests = [
    {"weight_kg": 5, "order_total": 4000, "city": " karachi "},
    {"weight_kg": 6, "order_total": 4000, "city": "Lahore"},
    {"weight_kg": 5, "order_total": 4000, "city": "Peshawar"},
    {"weight_kg": 6, "order_total": 4000, "city": "Peshawar"},
    {"weight_kg": 7, "order_total": 12000, "city": "Islamabad"}
]

for order in shipping_tests:
    weight_kg = order["weight_kg"]
    order_total = order["order_total"]
    city = order["city"].strip().title()

    if order_total > 10000:
        shipping_charge = 0
    elif city in ["Karachi", "Lahore", "Islamabad"]:
        if weight_kg <= 5:
            shipping_charge = 200
        else:
            shipping_charge = 400
    else:
        if weight_kg <= 5:
            shipping_charge = 350
        else:
            shipping_charge = 700

    print(f"Shipping to {city}: Rs {shipping_charge}")
```

### Question 5: Ticket triage

Read the task, then run the worked solution in the next cell.

```python
tickets = [
    {"plan": "enterprise", "hours_open": 6, "message": "Login is broken"},
    {"plan": "free", "hours_open": 1, "message": "I want a REFUND"},
    {"plan": "pro", "hours_open": 30, "message": "Dashboard is slow"},
    {"plan": "enterprise", "hours_open": 2, "message": "Question about invoices"},
    {"plan": "enterprise", "hours_open": 6, "message": "Please refund me"}
]

for ticket in tickets:
    plan = ticket["plan"].lower()
    hours_open = ticket["hours_open"]
    message = ticket["message"].lower()

    if plan == "enterprise" and hours_open > 4:
        priority = "Urgent"
    elif "refund" in message or "charged twice" in message or (
        plan == "pro" and hours_open > 24
    ):
        priority = "High"
    else:
        priority = "Normal"

    print(f"{plan}, {hours_open} hours: {priority}")

# Urgent is checked first. Therefore, an enterprise ticket open over
# 4 hours stays Urgent even if its message also mentions a refund.
```

# Demo 5: REPEATING WORK WITH LOOPS

A `for` loop repeats a block of code for each item in a sequence. You can use a variable to count items or accumulate a total. For example, start a total at `0`, then add each relevant value inside the loop. A loop can also contain an `if` statement to process only items that meet a condition.

---



### Question 1: Delivered revenue

Read the task, then run the worked solution in the next cell.

```python
orders = [
    {"id": "A101", "total": 4500, "status": "delivered"},
    {"id": "A102", "total": 12800, "status": "pending"},
    {"id": "A103", "total": 2300, "status": "delivered"}
]

delivered_revenue = 0

for order in orders:
    if order["status"] == "delivered":
        delivered_revenue += order["total"]

print(f"Delivered revenue: Rs {delivered_revenue:,}")
```

### Question 2: Numbered items

Read the task, then run the worked solution in the next cell.

```python
prices = [850, 1200, 400, 3100, 650]

counter = 0
total = 0
over_1000_count = 0

for price in prices:
    counter += 1
    total += price

    print(f"Item {counter}: Rs {price:,}")

    if price > 1000:
        over_1000_count += 1

print(f"Total: Rs {total:,}")
print(f"Items costing more than Rs 1,000: {over_1000_count}")
```

### Question 3: The biggest order

Read the task, then run the worked solution in the next cell.

```python
orders = [
    {"id": "A101", "total": 4500, "status": "delivered"},
    {"id": "A102", "total": 12800, "status": "pending"},
    {"id": "A103", "total": 2300, "status": "delivered"}
]

biggest_order = orders[0]

for order in orders:
    if order["total"] > biggest_order["total"]:
        biggest_order = order

print(f"Biggest order: {biggest_order['id']} (Rs {biggest_order['total']:,})")

# If two orders tie, this code keeps the first one it found.
```

### Question 4: Totals per customer

Read the task, then run the worked solution in the next cell.

```python
orders = [
    {"customer": "Noor Foods", "total": 4500},
    {"customer": "Karachi Crafts", "total": 12800},
    {"customer": "Noor Foods", "total": 2300},
    {"customer": "Bilal Traders", "total": 7600},
    {"customer": "Karachi Crafts", "total": 3900},
    {"customer": "Noor Foods", "total": 6100}
]

customer_totals = {}

for order in orders:
    customer = order["customer"]
    total = order["total"]

    if customer in customer_totals:
        customer_totals[customer] += total
    else:
        customer_totals[customer] = total

for customer, total in customer_totals.items():
    print(f"{customer}: Rs {total:,}")
```

### Question 5: Reorder report

Read the task, then run the worked solution in the next cell.

```python
inventory = [
    {"name": "Notebook", "stock": 8, "reorder_level": 15, "unit_cost": 90},
    {"name": "Pen", "stock": 140, "reorder_level": 50, "unit_cost": 25},
    {"name": "Stapler", "stock": 2, "reorder_level": 5, "unit_cost": 650},
    {"name": "Marker", "stock": 30, "reorder_level": 30, "unit_cost": 120},
    {"name": "Folder", "stock": 11, "reorder_level": 20, "unit_cost": 60}
]

total_cost = 0
products_to_reorder = 0

for product in inventory:
    if product["stock"] < product["reorder_level"]:
        quantity_to_order = (product["reorder_level"] * 2) - product["stock"]
        cost = quantity_to_order * product["unit_cost"]

        print(
            f"Reorder {quantity_to_order} x {product['name']} "
            f"(Rs {cost:,})"
        )

        total_cost += cost
        products_to_reorder += 1

if products_to_reorder == 0:
    print("All stock OK")
else:
    print(f"Total cost: Rs {total_cost:,}")

# Marker is not listed because its stock (30) is not below its
# reorder level (30); the question says below, not equal to.
```

# Demo 6: WRITING YOUR OWN FUNCTIONS

Define a function with `def`. Parameters are the names in the function definition; arguments are the values passed when calling it. `return` sends a result back to the caller. A parameter can have a default value, such as `rate=0.18`. Functions help give a piece of logic one clear name so it can be reused.

---



### Question 1: Formatting rupees

Read the task, then run the worked solution in the next cell.

```python
def format_rupees(amount):
    return f"Rs {amount:,.0f}"


print(format_rupees(12800))
print(format_rupees(999))
print(format_rupees(1250000))
print(f"Noor Foods owes {format_rupees(11520)}")
```

### Question 2: Return or print?

Read the task, then run the worked solution in the next cell.

```python
def add_tax(subtotal, rate=0.18):
    return subtotal + (subtotal * rate)


print(add_tax(7200))
print(add_tax(7200, 0.05))


def add_tax_with_print(subtotal, rate=0.18):
    result = subtotal + (subtotal * rate)
    print(result)


saved_result = add_tax_with_print(7200)
print(saved_result)

# The function prints the answer but does not return it.
# Therefore, saved_result contains None.
```

### Question 3: Shipping as a function

Read the task, then run the worked solution in the next cell.

```python
def shipping_charge(weight_kg, city, order_total=0):
    clean_city = city.strip().lower()

    if order_total > 10000:
        return 0

    if clean_city in ["karachi", "lahore", "islamabad"]:
        charge = 200
    else:
        charge = 350

    if weight_kg > 5:
        charge = charge * 2

    return charge


shipping_tests = [
    (5, " karachi ", 4000),
    (6, "Lahore", 4000),
    (5, "Peshawar", 4000),
    (6, "Peshawar", 4000),
    (7, "Islamabad", 12000)
]

for weight, city, order_total in shipping_tests:
    charge = shipping_charge(weight, city, order_total)
    print(f"Shipping to {city.strip().title()}: {format_rupees(charge)}")
```

### Question 4: Totals by status

Read the task, then run the worked solution in the next cell.

```python
orders = [
    {"id": "A101", "total": 4500, "status": "delivered"},
    {"id": "A102", "total": 12800, "status": "pending"},
    {"id": "A103", "total": 2300, "status": "delivered"}
]


def total_by_status(orders, status):
    total = 0

    for order in orders:
        if order["status"] == status:
            total += order["total"]

    return total


print(total_by_status(orders, "delivered"))
print(total_by_status(orders, "pending"))
print(total_by_status(orders, "cancelled"))
```

### Question 5: Refactor the reorder report

Read the task, then run the worked solution in the next cell.

```python
inventory = [
    {"name": "Notebook", "stock": 8, "reorder_level": 15, "unit_cost": 90},
    {"name": "Pen", "stock": 140, "reorder_level": 50, "unit_cost": 25},
    {"name": "Stapler", "stock": 2, "reorder_level": 5, "unit_cost": 650},
    {"name": "Marker", "stock": 30, "reorder_level": 30, "unit_cost": 120},
    {"name": "Folder", "stock": 11, "reorder_level": 20, "unit_cost": 60}
]


def needs_reorder(product):
    return product["stock"] < product["reorder_level"]


def reorder_quantity(product):
    return (product["reorder_level"] * 2) - product["stock"]


def reorder_cost(product):
    return reorder_quantity(product) * product["unit_cost"]


total_cost = 0
products_to_reorder = 0

for product in inventory:
    if needs_reorder(product):
        quantity = reorder_quantity(product)
        cost = reorder_cost(product)

        print(f"Reorder {quantity} x {product['name']} (Rs {cost:,})")

        total_cost += cost
        products_to_reorder += 1

if products_to_reorder == 0:
    print("All stock OK")
else:
    print(f"Total cost: Rs {total_cost:,}")

# Each function has one clear job, so its logic can be tested or changed
# without rewriting the whole report.
```
