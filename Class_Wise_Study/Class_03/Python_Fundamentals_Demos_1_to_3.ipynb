# Demo 1: VALUES, VARIABLES AND TYPES

A variable is a name that refers to a value. Use `=` to assign a value. Python values have types such as `str` (text), `int` (whole number), `float` (decimal number), and `bool` (`True` or `False`). Use `type()` to inspect a value's type. Arithmetic operators include `+`, `-`, `*`, `/`, `//` (floor division), and `%` (remainder).

---



### Question 1: Delivery charge

Read the task, then run the worked solution in the next cell.

```python
customer = "Karachi Crafts"
item = "lamp"
quantity = 3
price_per_lamp = 2400
delivery_charge = 350

order_total = quantity * price_per_lamp + delivery_charge

print(f"{customer}: {quantity} {item}(s)")
print(f"Order total including delivery: Rs {order_total:,}")
print(type(order_total))
```

### Question 2: Sales tax

Read the task, then run the worked solution in the next cell.

```python
subtotal = 7200
tax_rate = 0.18

tax_amount = subtotal * tax_rate
total_with_tax = subtotal + tax_amount

print(f"Tax amount: Rs {tax_amount}")
print(f"Total with tax: Rs {total_with_tax}")
print(type(tax_amount))

# The tax amount is a float because the tax rate contains a decimal.
```

### Question 3: Splitting a bill

Read the task, then run the worked solution in the next cell.

```python
bill = 4750
friends = 4

share = bill / friends
whole_rupees = bill // friends
left_over = bill % friends

print(f"Each person's exact share: Rs {share}")
print(f"Whole rupees per person: Rs {whole_rupees}")
print(f"Rupees left over: Rs {left_over}")
```

### Question 4: Discount and tax

Read the task, then run the worked solution in the next cell.

```python
phone_price = 15000
discount_rate = 0.10
tax_rate = 0.18

discounted_price = phone_price - (phone_price * discount_rate)
discount_then_tax = discounted_price + (discounted_price * tax_rate)

price_with_tax = phone_price + (phone_price * tax_rate)
tax_then_discount = price_with_tax - (price_with_tax * discount_rate)

print(f"Discount first, then tax: Rs {discount_then_tax:.2f}")
print(f"Tax first, then discount: Rs {tax_then_discount:.2f}")

# Both orders give the same result because both operations are percentages
# applied multiplicatively to the original price.
```

### Question 5: Packing boxes

Read the task, then run the worked solution in the next cell.

```python
items = 23
items_per_box = 6
shipping_per_box = 120

full_boxes = items // items_per_box
leftover_items = items % items_per_box

# A partly filled box also needs shipping.
total_boxes = full_boxes + (1 if leftover_items > 0 else 0)
shipping_cost = total_boxes * shipping_per_box

print(f"Full boxes: {full_boxes}")
print(f"Items left over: {leftover_items}")
print(f"Total boxes needed: {total_boxes}")
print(f"Shipping cost: Rs {shipping_cost}")

# Test with 24 items.
items = 24
full_boxes = items // items_per_box
leftover_items = items % items_per_box
total_boxes = full_boxes + (1 if leftover_items > 0 else 0)
shipping_cost = total_boxes * shipping_per_box

print(f"With 24 items, boxes needed: {total_boxes}")
print(f"Shipping cost: Rs {shipping_cost}")
```

# Demo 2: WORKING WITH TEXT

A string is text. String methods are called with a dot, such as `.strip()`, `.lower()`, `.title()`, `.replace()`, and `.upper()`. Use `len()` to count characters. An f-string starts with `f` and lets you place variables inside `{}`; formatting such as `:,.0f` adds comma separators and removes decimal places.

---



### Question 1: Clean up a name

Read the task, then run the worked solution in the next cell.

```python
raw_name = " aMNA rAZA "
name = raw_name.strip().title()

print(f"Welcome, {name}! Your name has {len(name)} characters.")
```

### Question 5: Usernames

Read the task, then run the worked solution in the next cell.

```python
first = " bilal "
last = "QURESHI "
city = "lahore"

first_name = first.strip().lower()
last_name = last.strip().lower()
city_part = city.strip().lower()[0:3]

username = f"{first_name}.{last_name}_{city_part}"

print(f"Welcome, {first_name.title()} {last_name.title()} from {city.strip().title()}!")
print(f"Your username is {username} ({len(username)} characters).")
```

# Demo 3: LISTS AND DICTIONARIES

A list stores several values in order. List positions start at `0`; use square brackets to access an item. Useful operations include `.append()`, `len()`, and `sum()`. A dictionary stores values under keys, like a spreadsheet row. Access a value with `dictionary[key]`, update it by assigning a new value, and use `.get()` when a key may be missing. A list of dictionaries is a common way to represent multiple records.

---



### Question 1: A product record

Read the task, then run the worked solution in the next cell.

```python
prices = [850, 1200, 400]

print(f"Last price: Rs {prices[-1]}")
print(f"Sum of prices: Rs {sum(prices)}")

product = {
    "name": "Notebook",
    "price": 850,
    "in_stock": True
}

product["in_stock"] = False
print(product["name"])
```

### Question 2: A week of sales

Read the task, then run the worked solution in the next cell.

```python
daily_sales = [18500, 22300, 19800, 25400, 31200, 42800, 38600]

weekly_total = sum(daily_sales)
daily_average = weekly_total / len(daily_sales)

print(f"Monday sales: Rs {daily_sales[0]:,.0f}")
print(f"Sunday sales: Rs {daily_sales[6]:,.0f}")
print(f"Weekly total: Rs {weekly_total:,.0f}")
print(f"Daily average: Rs {daily_average:,.0f}")
print(f"Best day's sales: Rs {max(daily_sales):,.0f}")
print(f"Worst day's sales: Rs {min(daily_sales):,.0f}")
```

### Question 4: A list of orders

Read the task, then run the worked solution in the next cell.

```python
orders = [
    {"id": "A101", "customer": "Noor Foods", "total": 4500},
    {"id": "A102", "customer": "Karachi Crafts", "total": 12800},
    {"id": "A103", "customer": "Bilal Traders", "total": 7600}
]

print(f"Second order customer: {orders[1]['customer']}")

orders[2]["total"] = orders[2]["total"] - 500

orders.append({
    "id": "A104",
    "customer": "City Store",
    "total": 9200
})

print(
    f"{len(orders)} orders. "
    f"The first is {orders[0]['id']} and the last is {orders[-1]['id']}."
)
```

### Question 5: A customer profile

Read the task, then run the worked solution in the next cell.

```python
customer = {
    "name": "Noor Foods",
    "address": {"city": "Karachi", "area": "Clifton"},
    "order_totals": [4500, 12800, 2300]
}

print(f"City: {customer['address']['city']}")
print(f"Number of orders: {len(customer['order_totals'])}")
print(f"Total spent: Rs {sum(customer['order_totals']):,}")

customer["order_totals"].append(6100)
print(f"New total spent: Rs {sum(customer['order_totals']):,}")

# Accessing customer["phone"] directly would raise a KeyError if the key
# does not exist, so get() provides a safe default value.
print(customer.get("phone", "No phone number"))
```
